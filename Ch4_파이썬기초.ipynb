### 리스트 메서드

```python
# append(x)리스트에 항목 1개 추가
a = [1, 2]
a.append(3)

# extend(iterable)여러 항목을 한꺼번에 추가
a = [1, 2]
a.extend([3, 4])

# pop(i)항목을 꺼내면서 삭제
a = [1, 2, 3]
val = a.pop()
print(val)

# sort()리스트 정렬
a = [3, 1, 2]
a.sort()
print(a)

# remove(x)특정 항목 삭제
a = [1, 2, 3, 2]
a.remove(2)
print(a)

# index(x)특정 항목의 위치 찾기
a = ['a', 'b', 'c']
idx = a.index('b')
print(idx)

# copy()리스트 복사
a = [1, 2]
b = a.copy()
print(b)

# 지정한 인덱스 i 위치에 항목 x를 삽입
a = [1, 3]
a.insert(1, 2)
print(a)

# insert(i, x)특정 위치에 항목 삽입
a = ['a', 'b', 'c']
idx = a.index('b')
print(idx)

# reverse()순서 뒤집기
a = [1, 2, 3]
a.reverse()
print(a)

# count(x)특정 항목의 개수 세기
a = [1, 2, 2, 3]
cnt = a.count(2)
print(cnt)

# clear()
a = [1, 2, 3]
a.clear()
print(a)
```

### 가변 인자 (*args, **kwargs)

```python
# 위치 가변인자 (*args)
def sum_all(*args):
    # 여러 개의 인자를 튜플로 묶음
    return sum(args)

print(sum_all(10, 20))
print(sum_all(1, 2, 3, 4, 5))
```

```python
# 키워드 가변인자 (**kwargs)   ※ kwargs: Keyword Arguments의 약자
def print_user_info(**kwargs): 
# kwargs는 {'name': 'Gemini', 'age': 20} 처럼 딕셔너리로 묶입니다.
   for key, value in kwargs.items():
     print(f"{key}: {value}") 

print_user_info(name="Gemini", version="3.0")
```

### 변수의 범위: 전역, 지역

```python
global_var = 100 # 전역 변수

def my_func():
    local_var = 20 # 지역 변수
    print(global_var) # 읽기 가능

my_func()
# print(local_var) # 에러! (함수 밖에서는 지역 변수가 안 보임)
```

```python
# 함수 안에서 전역 변수를 수정하려면 global 키워드가 필요하지만, 남용하면 코드가 꼬일 수 있어 주의해야 합니다.
count = 0   # 전역 변수

def increase():
    global count
    count = count + 1

increase()
print(count)
```

```python
import pandas as pd 

# 1. 샘플 데이터 생성 
data = { 'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'], 'Score': [95, 82, 75, 89, 60] } 
df = pd.DataFrame(data) 

# 2. apply와 lambda를 사용하여 등급(Grade) 계산 
# 로직: 90점 이상 'A', 아니면 (80점 이상 'B', 아니면 'C’) 
df['Grade'] = df['Score'].apply( lambda x: 'A' if x >= 90 else ('B' if x >= 80 else 'C') )

 # 3. 결과 출력 
print("--- 성적 결과 데이터프레임 ---")
print(df)
```

# 8) 클래스

### 클래스와 객체

```python
class FishBread:    
    # 생성자: 붕어빵이 태어날 때 실행됨
    def __init__(self, f, p):
        self.flavor = f  # 맛 (속성)
        self.price = p   # 가격 (속성)    
    # 메서드: 붕어빵의 행동
    def show_info(self):  
        # 여기서 self는 호출한 녀석(bread1 또는 bread2)이 됩니다.
        print(f"{self.flavor} 붕어빵은 {self.price}원입니다.")

# 객체 생성 (틀로 빵 찍어내기)
bread1 = FishBread("팥", 1000)      # self가 bread1이 되어 '팥'을 저장
bread2 = FishBread("슈크림", 1200)  # self가 bread2가 되어 '슈크림'을 저장
bread1.show_info() # bread1의 정보 출력
bread2.show_info() # bread2의 정보 출력
```

### 클래스 변수 vs 인스턴스 변수

```python
class Monster:
    count = 0  # [클래스 변수] 모든 몬스터가 공유 (총 몬스터 수)

    def __init__(self, name):
        self.name = name  # [인스턴스 변수] 몬스터마다 다름 (내 이름)
        Monster.count += 1 # 몬스터가 태어날 때마다 전체 카운트 증가

m1 = Monster("슬라임")
print(f"현재 몬스터 수: {Monster.count}마리") # 1마리

m2 = Monster("골렘")
print(f"현재 몬스터 수: {Monster.count}마리") # 2마리 (m1에게도 영향을 줌)
```

### 상속

```python
# FishBread의 기능을 물려받음
class PremiumFishBread(FishBread):
    def discount(self):
        self.price -= 100 # 100원 할인 기능 추가

p_bread = PremiumFishBread("황금", 2000)
p_bread.discount()
p_bread.show_info() # 황금 붕어빵은 1900원입니다.
```

### 전기차동차 생산

```python
class Car:
    def __init__(self, model):
        self.model = model
        self.speed = 0

    def accelerate(self, value):
        self.speed += value
        print(f"{self.model} 가속! 현재 속도: {self.speed}km/h")

    def brake(self, value):
        self.speed -= value
        if self.speed < 0:
            self.speed = 0
        print(f"{self.model} 감속! 현재 속도: {self.speed}km/h")
my_car = Car("테슬라")
my_car.accelerate(50)
my_car.brake(20)
```
